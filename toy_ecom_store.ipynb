```python
from sqlalchemy import create_engine
```

### Export data to sql database

```python
engine = create_engine("mysql+pymysql://root:1510@localhost:3306/toys_ecommerce_dataset")
```

```python
# Function to store dataframe in MySQL
def store_dataframe(df, table_name, if_exists="replace"):
    df.to_sql(table_name, engine, if_exists="replace", index=False)
```

```python
store_dataframe(orders,"orders")
store_dataframe(order_items,"order_items")
store_dataframe(refunds,"refunds")
store_dataframe(products,"products")
store_dataframe(website_pageviews,"website_pageviews")
```

```python
store_dataframe(website_sessions,"website_sessions")
```
